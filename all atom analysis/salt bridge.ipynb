```python
import MDAnalysis as mda
import numpy as np

# 定义计算盐桥的函数
def saltbridge_Kpho(u, distance_threshold=4.0,start=0,end = None):
    # 加载轨迹和拓扑文件
    # 选择LYS的氮原子（NZ）和GLU的负电荷氧原子（OE1, OE2）
    lys_nitrogens = u.select_atoms("resname LYS and name NZ")  # LYS的氮原子
    glu_oxygens = u.select_atoms("resname SER and (name O1P or name O2P or name OT)")  # GLU的负电荷氧原子
    
    # 用于保存每帧中形成盐桥的数量
    salt_bridge_counts = []
    
    # 遍历每一帧
    for ts in u.trajectory[start:end ]:
        count = 0
        # 计算LYS氮原子和GLU氧原子之间的距离
        for lys in lys_nitrogens:
            for glu in glu_oxygens:
                # 计算LYS氮原子和GLU氧原子之间的距离
                distance = np.linalg.norm(lys.position - glu.position)
                if distance < distance_threshold:
                    count += 1
                    # 如果找到一对满足条件的氮-氧原子，可以直接跳出循环
                    break
        # 将每帧的盐桥数添加到列表
        salt_bridge_counts.append(count)
    return salt_bridge_counts

def saltbridge_OHpho(u, distance_threshold=4.0,start=0,end = None):
    # 加载轨迹和拓扑文件
    # 选择LYS的氮原子（NZ）和GLU的负电荷氧原子（OE1, OE2）
    lys_nitrogens = u.select_atoms("resname THR and name OG1")  # LYS的氮原子
    glu_oxygens = u.select_atoms("resname SER and (name O1P or name O2P or name OT)")  # GLU的负电荷氧原子
    
    # 用于保存每帧中形成盐桥的数量
    salt_bridge_counts = []
    
    # 遍历每一帧
    for ts in u.trajectory[start:end ]:
        count = 0
        # 计算LYS氮原子和GLU氧原子之间的距离
        for lys in lys_nitrogens:
            for glu in glu_oxygens:
                # 计算LYS氮原子和GLU氧原子之间的距离
                distance = np.linalg.norm(lys.position - glu.position)
                if distance < distance_threshold:
                    count += 1
                    # 如果找到一对满足条件的氮-氧原子，可以直接跳出循环
                    break
        # 将每帧的盐桥数添加到列表
        salt_bridge_counts.append(count)
    return salt_bridge_counts

def saltbridge_Nterpho(u, distance_threshold=4.0,start=0,end = None):
    # 加载轨迹和拓扑文件
    # 选择LYS的氮原子（NZ）和GLU的负电荷氧原子（OE1, OE2）
    lys_nitrogens = u.select_atoms("resid 1 and name N")  # LYS的氮原子
    glu_oxygens = u.select_atoms("resname SER and (name O1P or name O2P or name OT)")  # GLU的负电荷氧原子
    
    # 用于保存每帧中形成盐桥的数量
    salt_bridge_counts = []
    
    # 遍历每一帧
    for ts in u.trajectory[start:end ]:
        count = 0
        # 计算LYS氮原子和GLU氧原子之间的距离
        for lys in lys_nitrogens:
            for glu in glu_oxygens:
                # 计算LYS氮原子和GLU氧原子之间的距离
                distance = np.linalg.norm(lys.position - glu.position)
                if distance < distance_threshold:
                    count += 1
                    # 如果找到一对满足条件的氮-氧原子，可以直接跳出循环
                    break
        # 将每帧的盐桥数添加到列表
        salt_bridge_counts.append(count)
    return salt_bridge_counts
```
