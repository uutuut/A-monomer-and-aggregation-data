```python
import numpy as np
import matplotlib.pyplot as plt

# 定义新温度
new_temperatures_p = np.array([310,327.98,347.002,367.128,388.421,410.949,434.783,460])
replica_temp = np.loadtxt("E:/a40/a40/conv/replica_temp.xvg")

# 替换replica_temp[:, 1]中等于0-41的值
for i in range(len(new_temperatures_p)):
    replica_temp[replica_temp[:, 1] == i, 1] = new_temperatures_p[i]

# 可视化更新的数据
plt.figure(figsize=(10, 7.5), dpi=600)  # 高分辨率
plt.plot(replica_temp[:, 0]/1000, replica_temp[:, 1], color='black', linewidth=2)

# 设置y轴刻度
plt.yticks(fontsize=25)
plt.ylim(310,460)
plt.xticks(fontsize=25)
plt.xlim(0,700)

# 设置标签和标题
plt.xlabel('Time (ns)', fontsize=30)
plt.ylabel('Temperature (K)', fontsize=30)
plt.tick_params(axis='both', which='both', direction='out', length=10, width=2)
# 设置图例
#plt.legend(fontsize=12, loc='upper right', frameon=False)

# 格式化网格和背景
plt.grid(False)
plt.gca().set_facecolor('white')

# 自适应布局并保存图像
plt.tight_layout()
plt.savefig("E:/a40/temp_evo_a40.png", bbox_inches='tight')  # 保存图像
plt.show()  # 显示图形
```
