```python
import numpy as np
import matplotlib.pyplot as plt

# 定义读取数据的函数


data_100_500ns_1 = np.loadtxt('E:/a40/a40/asphericity.txt')
data_100_500ns_2 = np.loadtxt('E:/a40/a40p8/asphericity.txt')
data_100_500ns_3 = np.loadtxt('E:/a40/a40p26/asphericity.txt')
# 计算三个数据的概率分布
hist_1, bin_edges_1 = np.histogram(data_100_500ns_1, bins=50, density=False)
hist_2, bin_edges_2 = np.histogram(data_100_500ns_2, bins=50, density=False)
hist_3, bin_edges_3 = np.histogram(data_100_500ns_3, bins=50, density=False)

# 归一化概率分布
prob_1 = hist_1 / np.sum(hist_1)
prob_2 = hist_2 / np.sum(hist_2)
prob_3 = hist_3 / np.sum(hist_3)

# 计算每个bin的中心值
bin_centers_1 = (bin_edges_1[:-1] + bin_edges_1[1:]) / 2
bin_centers_2 = (bin_edges_2[:-1] + bin_edges_2[1:]) / 2
bin_centers_3 = (bin_edges_3[:-1] + bin_edges_3[1:]) / 2

# 绘制概率分布
plt.figure(figsize=(8, 10), dpi=600)

# 绘制三个区间的数据的概率分布
plt.plot(bin_centers_1, prob_1 * 100, label="Aβ40", color="blue", linewidth=2)
plt.plot(bin_centers_2, prob_2 * 100, label="Aβ40p8", color="green", linewidth=2)
plt.plot(bin_centers_3, prob_3 * 100, label="Aβ40p26", color="red", linewidth=2)

# 设置标签和标题
plt.xlabel('Asphericity', fontsize=35)
plt.ylabel('Probability(%)', fontsize=35)
plt.ylim(0,)
# 添加图例
plt.legend(fontsize=30, frameon=False)

# 格式化网格和刻度
plt.grid(False)
plt.xticks(fontsize=30)
plt.yticks(fontsize=30)
plt.tick_params(axis='both', which='both', direction='out', length=10, width=2)
# 显示图形
plt.tight_layout()
plt.savefig("E:/a40/Asphericity.png")
plt.show()
```
